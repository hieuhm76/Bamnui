```python

from IPython.display import display, HTML
import base64

game = r"""
<!DOCTYPE html>
<html>
<head>
<meta charset="UTF-8">
<style>
body {
  background:#172033; color:white; font-family:Arial;
  text-align:center; padding:20px;
}
#clock {font-size:100px;font-weight:bold;margin:25px;}
button {
  padding:15px 28px;font-size:20px;border:0;
  border-radius:12px;color:white;margin:5px;cursor:pointer;
}
#start {background:#16a34a}
#stop {background:#64748b}
#reset {background:#475569;font-size:16px}
</style>
</head>
<body>
<h2>Bấm nút phát biểu</h2>
<p>Bấm STOP sát thời điểm đủ 10 giây</p>
<div id="clock">10</div>
<button id="start">START</button>
<button id="stop" disabled>STOP</button>
<h2 id="result">Sẵn sàng?</h2>
<div id="detail"></div>
<p id="record">Chưa có thành tích</p>
<button id="reset">CHƠI LẠI</button>

<script>
let startTime=0, timer=null, running=false;
let attempts=0, best=null;

const clock=document.getElementById('clock');
const start=document.getElementById('start');
const stop=document.getElementById('stop');
const reset=document.getElementById('reset');
const result=document.getElementById('result');
const detail=document.getElementById('detail');
const record=document.getElementById('record');

function update(){
  if(!running) return;
  const elapsed=(performance.now()-startTime)/1000;
  const remaining=Math.max(0,10-Math.floor(elapsed));
  clock.textContent=String(remaining).padStart(2,'0');
  clock.style.color=remaining===0?'#f97316':'white';
}

start.onclick=function(){
  if(running) return;
  clearInterval(timer);
  running=true;
  startTime=performance.now();
  clock.textContent='10';
  clock.style.color='white';
  result.textContent='ĐANG ĐẾM...';
  result.style.color='white';
  detail.textContent='';
  start.disabled=true;
  stop.disabled=false;
  stop.style.background='#f97316';
  timer=setInterval(update,10);
};

stop.onclick=function(){
  if(!running) return;
  const elapsed=(performance.now()-startTime)/1000;
  running=false;
  clearInterval(timer);
  attempts++;

  const error=elapsed-10;
  const absError=Math.abs(error);
  let color,message;

  if(error<0){
    color='#dc2626'; message='BẤM QUÁ SỚM!';
  }else if(error<=0.20){
    color='#16a34a'; message='XUẤT SẮC!';
  }else if(error<=0.50){
    color='#2563eb'; message='KHÁ TỐT!';
  }else if(error<=0.80){
    color='#eab308'; message='HƠI CHẬM!';
  }else{
    color='#dc2626'; message='QUÁ CHẬM!';
  }

  clock.textContent=String(
    Math.max(0,10-Math.floor(elapsed))
  ).padStart(2,'0');

  stop.style.background=color;
  stop.disabled=true;
  result.textContent=message;
  result.style.color=color;

  detail.innerHTML=
    'Thời gian thực tế: <b>'+elapsed.toFixed(3)+
    ' giây</b><br>Sai lệch: <b>'+
    (error>0?'+':'')+error.toFixed(3)+' giây</b>';

  if(best===null || absError<best) best=absError;
  record.innerHTML='Số lần thử: '+attempts+
    '<br>Sai lệch tốt nhất: '+best.toFixed(3)+' giây';

  start.disabled=false;
  start.textContent='THỬ LẠI';
};

reset.onclick=function(){
  clearInterval(timer);
  running=false;
  clock.textContent='10';
  clock.style.color='white';
  result.textContent='Sẵn sàng?';
  result.style.color='white';
  detail.textContent='';
  start.disabled=false;
  start.textContent='START';
  stop.disabled=true;
  stop.style.background='#64748b';
};
</script>
</body>
</html>
"""

encoded = base64.b64encode(game.encode("utf-8")).decode("ascii")

display(HTML(
    '<iframe sandbox="allow-scripts" '
    'src="data:text/html;base64,' + encoded + '" '
    'width="100%" height="520" '
    'style="border:0;border-radius:16px;"></iframe>'
))
```
